```python
import matplotlib.pyplot as plt

def main():
    # Array untuk menyimpan data
    nilai_mahasiswa = []

    print("=== Sistem Pengelolaan Nilai Mahasiswa ===")

    # 1. Input 10 nilai
    for i in range(10):
        while True:
            try:
                nilai = float(input(f"Masukkan nilai mahasiswa ke-{i+1}: "))
                if 0 <= nilai <= 100:
                    nilai_mahasiswa.append(nilai)
                    break
                else:
                    print("Error: Nilai harus berada di rentang 0 - 100.")
            except ValueError:
                print("Error: Input tidak valid. Masukkan angka.")

    # 2. Tampilkan nilai tertinggi & terendah
    nilai_tertinggi = max(nilai_mahasiswa)
    nilai_terendah = min(nilai_mahasiswa)

    print("\n=== Hasil Analisis ===")
    print(f"Nilai Tertinggi : {nilai_tertinggi}")
    print(f"Nilai Terendah  : {nilai_terendah}")

    # 3. Hitung rata-rata
    rata_rata = sum(nilai_mahasiswa) / len(nilai_mahasiswa)
    print(f"Rata-rata Nilai : {rata_rata:.2f}")

    # 4. Hitung jumlah mahasiswa lulus (>= 60)
    mahasiswa_lulus = sum(1 for n in nilai_mahasiswa if n >= 60)
    mahasiswa_gagal = len(nilai_mahasiswa) - mahasiswa_lulus
    print(f"Jumlah Lulus    : {mahasiswa_lulus} mahasiswa")
    print(f"Jumlah Gagal    : {mahasiswa_gagal} mahasiswa")

    # Siapkan Canvas untuk Grafik
    plt.figure(figsize=(12, 5))

    # 5. Buat grafik untuk menggambarkan nilai tertinggi dan terendah
    plt.subplot(1, 2, 1) # Baris 1, Kolom 2, Grafik ke-1
    kategori_minmax = ['Tertinggi', 'Terendah']
    nilai_minmax = [nilai_tertinggi, nilai_terendah]
    plt.bar(kategori_minmax, nilai_minmax, color=['#4CAF50', '#F44336'])
    plt.title('Grafik Nilai Tertinggi vs Terendah')
    plt.ylabel('Nilai')

    # 6. Buat grafik untuk menggambarkan data kelulusan
    plt.subplot(1, 2, 2) # Baris 1, Kolom 2, Grafik ke-2
    kategori_status = ['Lulus', 'Tidak Lulus']
    jumlah_status = [mahasiswa_lulus, mahasiswa_gagal]
    plt.pie(jumlah_status, labels=kategori_status, autopct='%1.1f%%', colors=['#2196F3', '#FFC107'], startangle=90)
    plt.title('Persentase Status Kelulusan')

    # Tampilkan grafik
    plt.tight_layout()
    plt.show()

if __name__ == "__main__":
    main()1
```
